## Ejercicio 1.4: Resumen sobre NVIDIA Cosmos

**NVIDIA Cosmos** es una plataforma de **IA Física** (*Physical AI*) desarrollada para acelerar la creación de robots y vehículos autónomos[cite: 1]. Sus pilares clave son:

1. **Modelos de Base del Mundo (World Foundation Models):** Permiten a los sistemas físicos simular, predecir y entender entornos reales con alta precisión.
2. **Simulación en Omniverse:** Facilita el entrenamiento en entornos virtuales realistas procesando visión, física y lenguaje en tiempo real.
3. **Conexión entre lo Digital y lo Físico:** Cierra la brecha entre el procesamiento de software y la acción mecánica en el mundo real[cite: 1].
